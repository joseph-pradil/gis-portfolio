```python
import geopandas as gpd
from shapely.geometry import Point
import matplotlib.pyplot as plt

# Create some sample points (pretend these are field survey locations)
points_data = {
    'location': ['Site A', 'Site B', 'Site C', 'Site D', 'Site E'],
    'value': [10, 25, 15, 30, 20],
    'geometry': [
        Point(151.2093, -33.8688),  # Sydney
        Point(144.9631, -37.8136),  # Melbourne
        Point(153.0251, -27.4698),  # Brisbane
        Point(115.8605, -31.9505),  # Perth
        Point(138.6007, -34.9285),  # Adelaide
    ]
}

# Create GeoDataFrame
gdf = gpd.GeoDataFrame(points_data, crs="EPSG:4326")

# Plot it
fig, ax = plt.subplots(figsize=(10, 8))
gdf.plot(ax=ax, color='red', markersize=100, alpha=0.7)

# Add labels
for idx, row in gdf.iterrows():
    ax.annotate(row['location'], 
                xy=(row.geometry.x, row.geometry.y),
                xytext=(3, 3), textcoords="offset points",
                fontsize=12)

ax.set_title('Australian Survey Sites', fontsize=16)
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
plt.tight_layout()
plt.savefig('australian_sites.png', dpi=150)
plt.show()

print("Script 1 complete! Points plotted over Australia.")
print(gdf)
```
